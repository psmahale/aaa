```
titanic = sns.load_dataset('titanic')

plt.figure(figsize=(10,6))
sns.histplot(data=titanic, x='fare', kde=True, bins=30, color='blue')
plt.title('Price of Fare')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')
plt.show()
```
