```
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, MinMaxScaler
import matplotlib.pyplot as plt
import seaborn as sns
```

```
# Load the dataset from UCI repository
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
col_names = ['Sepal_Length','Sepal_Width','Petal_Length','Petal_Width','Species']
df = pd.read_csv(url, names=col_names)
df.head()
```

```
df.info()
df.describe(include='all')
df.isnull().sum()
```

```
# Check data types
print(df.dtypes)

# Normalize numeric columns
scaler = MinMaxScaler()
x = df.iloc[:, :4]
x_scaled = scaler.fit_transform(x)
df_normalized = pd.DataFrame(x_scaled, columns=col_names[:4])
df_normalized.head()
```

```
# Encode the Species column using Label Encoding
le = LabelEncoder()
df['Species'] = le.fit_transform(df['Species'])
df.head()
```

```
# Apply One Hot Encoding
ohe = OneHotEncoder(sparse=False)
encoded_species = ohe.fit_transform(df[['Species']])
encoded_df = pd.DataFrame(encoded_species, columns=['Species_0', 'Species_1', 'Species_2'])
final_df = pd.concat([df.drop('Species', axis=1), encoded_df], axis=1)
final_df.head()
```

```
# Apply Dummy Variable Encoding
dummy_df = pd.get_dummies(df, columns=['Species'], prefix='Species', drop_first=True)
dummy_df.head()
```
